### Cluster and Outlier (Local Moran's I) Analysis results

In ArcGIS I ran the Cluster and Outlier tool to examine the relationship between the percentage of chinese people and restaurants per census tract across the urban/rural gradient.  In this notebook I examine the results and prepare the data for charting.

```python
import pandas as pd
```

```python
# load cluster/outlier results

clusters = pd.read_csv('ClusterOutlier_with_locales_ExportTable.csv')
print(clusters.shape)
clusters.head(-10)
```

```python
# determine STATEFP data type
clusters['STATEFP'].dtype
```

```python
# get rows where STATEFP is null
clusters[clusters['STATEFP'].isnull()]
```

```python
# remove rows with null STATEFP
clusters = clusters[~clusters['STATEFP'].isnull()]
```

```python
# change STATEFP to integers
clusters['STATEFP'] = clusters['STATEFP'].astype(int)
clusters.head()
```

```python
# count rows where STATEFP is 72
clusters[clusters['STATEFP'] == 72].shape

```

```python
# remove Puerto Rico for now
clusters = clusters[clusters['STATEFP'] != 72]
clusters.shape
```

```python
# change Loc_CODE to integers
clusters['Loc_CODE'] = clusters['Loc_CODE'].astype(int)
clusters.head()
```

```python
# check if any rows are not assigned to a locale
clusters[clusters['Loc_CODE'].isnull()].shape
```

```python
# number of tracts by cluster and locale type
clusters_grouped = clusters.groupby(['COType', 'Loc_CODE', 'LOC_NAME']).size().reset_index(name='count')
clusters_grouped['percentage'] = clusters_grouped['count'] / clusters_grouped['count'].sum() * 100
clusters_grouped

```

```python
# save as csv for charting
clusters_grouped.to_csv('clusters_grouped.csv', index=False)
```

```python
# looking for proof that restaurant-to-chinese-population ratios are higher on average in small towns

from scipy import stats

urban_ratios = clusters[clusters['Loc_CODE'].isin([11, 12, 13])]['rest_per_100ch'].tolist()
suburban_ratios = clusters[clusters['Loc_CODE'].isin([21, 22, 23])]['rest_per_100ch'].tolist()
smalltown_ratios = clusters[clusters['Loc_CODE'].isin([31, 32, 33])]['rest_per_100ch'].tolist()
rural_ratios = clusters[clusters['Loc_CODE'].isin([41, 42, 43])]['rest_per_100ch'].tolist()

t_stat, p_value = stats.ttest_ind(urban_ratios, smalltown_ratios)
```

```python
p_value
```

```python
t_stat
```

```python
# calculate the mean of each list
urban_mean = sum(urban_ratios) / len(urban_ratios)
smalltown_mean = sum(smalltown_ratios) / len(smalltown_ratios)

# print the means
print(f"urban ratios mean: {urban_mean}")
print(f"smalltown ratios mean: {smalltown_mean}")
```
