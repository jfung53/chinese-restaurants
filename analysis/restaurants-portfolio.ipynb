```python
import pandas as pd
```

```python
# restaurant data with locale type
restaurants = pd.read_csv('restaurants-us.csv')
print(restaurants.shape)
restaurants.head()
```

```python
# Change data type for postcode, LOCALE, and STATEFP to integers
restaurants['postcode'] = restaurants['postcode'].astype('Int64')
restaurants['LOCALE'] = restaurants['LOCALE'].astype('Int64')
restaurants['STATEFP'] = restaurants['STATEFP'].astype('Int64')
```

```python
restaurants.head()
```

```python
# fips codes table
fips = pd.read_csv('fips-codes.csv')
print(fips.shape)
fips.head()
```

```python
# Change STATEFP data type to integers
fips['STATEFP'] = fips['STATEFP'].astype(int)
```

```python
# Fill missing 'region' (state abbreviation) values in restaurants using STATEFP and fips 'ABB'
mask = restaurants['region'].isnull() | (restaurants['region'] == '')
restaurants.loc[mask, 'region'] = restaurants.loc[mask, 'STATEFP'].map(
    fips.set_index('STATEFP')['ABB']
)
```

```python
missing_region_df = restaurants[restaurants['region'].isnull() | (restaurants['region'] == "")]
missing_region_df.head()
```

```python
missing_region_df.shape
```

```python
restaurants.to_csv("restaurants-us-cleaned.csv", index=False)
```
