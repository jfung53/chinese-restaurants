### Chinese Restaurant Name Analysis
---

Do small town Chinese restaurants tend to be named differently from the ones in big cities?

```python
import pandas as pd
```

```python
# load chinese restaurants by locale type

restaurants = pd.read_csv('us_restaurants_with_locales.csv')
print(restaurants.shape)
restaurants.head()
```

```python
# remove all rows with no locale
restaurants = restaurants.dropna(subset=['LOCALE'])
```

```python
# remove decimal place
restaurants['LOCALE'] = restaurants['LOCALE'].astype(int)
```

```python

locale_counts = restaurants['LOC_NAME'].value_counts().sort_values(ascending=False)
locale_counts
```

```python
# restaurants in cities
cities = restaurants[restaurants['LOCALE'].isin([11, 12, 13])].copy()
cities.shape
```

```python
# restaurants in suburbs
suburbs = restaurants[restaurants['LOCALE'].isin([21, 22, 23])].copy()
suburbs.shape
```

```python
# restaurants in small towns
towns = restaurants[restaurants['LOCALE'].isin([31, 32, 33])].copy()
towns.shape
```

```python
# restaurants in rural areas
rural = restaurants[restaurants['LOCALE'].isin([41, 42, 43])].copy()
rural.shape
```

```python
city_name_counts = cities['name'].value_counts()
print(city_name_counts.shape)
city_name_counts
```
